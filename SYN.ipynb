### Unified Trilingual Document Engine (INV + EPO + FSD)
This module handles the "One Action" generation for all three syndicate documents. It ensures:
1. **Minimalist Futuristic Tone:** Removes 'chit-chat' bureaucratic text.
2. **Trilingual Headers:** Arabic, Kurmanji, and English support.
3. **Branch Isolation:** Logic is tied to the city code (HAS/QAM/DER).

```python
import pandas as pd
import datetime

class SyndicateDocEngine:
    def __init__(self, branch_code):
        self.branch_code = branch_code
        self.branch_registry = {
            'HAS': {'ar': 'فرع الحسكة', 'ku': 'Tayê Hesekê', 'en': 'Al Hasakah Branch', 'accent': '#0056b3'},
            'QAM': {'ar': 'فرع القامشلي', 'ku': 'Tayê Qamişlo', 'en': 'Qamishli Branch', 'accent': '#d9534f'},
            'DER': {'ar': 'فرع ديريك', 'ku': 'Tayê Dêrikê', 'en': 'Derik Branch', 'accent': '#5cb85c'}
        }

    def get_trilingual_labels(self):
        return {
            'INV': {'ar': 'فاتورة', 'ku': 'Fatûre', 'en': 'Invoice'},
            'EPO': {'ar': 'أمر صرف', 'ku': 'Fermana Peredanê', 'en': 'Payout Order'},
            'FSD': {'ar': 'وثيقة مالية', 'ku': 'Belgeya Darayî', 'en': 'Financial Document'}
        }

    def generate_unified_pack(self, data_payload):
        """Generates metadata for all 3 documents in one action."""
        branch = self.branch_registry.get(self.branch_code)
        labels = self.get_trilingual_labels()
        timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M')

        pack = {}
        for doc_type in ['INV', 'EPO', 'FSD']:
            pack[doc_type] = {
                'header': branch,
                'title': labels[doc_type],
                'data': data_payload,
                'footer': f"Verified by Central Accumulator | {timestamp}",
                'qr': f"VERIFY-{self.branch_code}-{doc_type}-2026"
            }

        return pack

# Implementation Demo
engine = SyndicateDocEngine('HAS')
sample_transaction = {'engineer': 'Jihad H.', 'amount': '5,000,000 SYP', 'model': 1}
doc_pack = engine.generate_unified_pack(sample_transaction)

print("UNIFIED BUNDLE GENERATED (Trilingual/2026 Ready):")
for doc, content in doc_pack.items():
    print(f"- {doc}: {content['title']['ar']} | {content['title']['ku']} | {content['title']['en']}")
```

### Accounting Model Integration & Data Population
We are now mapping the trilingual interface to the 7 accounting models defined in `08_central_accumulator-v2.sql`.

**Logic Flow:**
- **Model 1-5:** Handling specific stage-based deductions.
- **Model 6-7:** Handling mutual fund and administrative fee structures.
- **Precision:** All calculations remain in Syrian Pounds (SYP) with elite minimalist formatting.

```python
import pandas as pd

def get_model_calculation(amount, model_id, exchange_rate=13500):
    """
    Core Syndicate Accounting Logic (v2026).
    Only Model 7 (GS) assumes USD input converted to SYP.
    Models 1-6 are treated as standard SYP operations unless specified.
    """
    models = {
        1: {'label': 'Standard Project', 'rate': 0.05, 'is_usd': False},
        2: {'label': 'Consultancy', 'rate': 0.07, 'is_usd': False},
        3: {'label': 'Supervision', 'rate': 0.04, 'is_usd': False},
        4: {'label': 'Technical Report', 'rate': 0.03, 'is_usd': False},
        5: {'label': 'Surveying', 'rate': 0.06, 'is_usd': False},
        6: {'label': 'Administrative', 'rate': 0.02, 'is_usd': False},
        7: {'label': 'Syndicate Fund (GS)', 'rate': 0.10, 'is_usd': True}
    }

    config = models.get(model_id, models[1])

    if config['is_usd']:
        gross_usd = amount
        gross_syp = amount * exchange_rate
        deduction_syp = gross_syp * config['rate']
        net_payout = gross_syp - deduction_syp
    else:
        gross_usd = None
        gross_syp = amount
        deduction_syp = amount * config['rate']
        net_payout = amount - deduction_syp

    return {
        'model_name': config['label'],
        'is_mixed': config['is_usd'],
        'gross_usd': gross_usd,
        'gross_syp': gross_syp,
        'deductions': deduction_syp,
        'net_payout': net_payout,
        'exchange_rate': exchange_rate if config['is_usd'] else None
    }

# Audit mixed-currency (Model 7)
gs_audit = get_model_calculation(amount=1500, model_id=7, exchange_rate=14200)
print(f"GS AUDIT: ${gs_audit['gross_usd']} @ {gs_audit['exchange_rate']} -> {gs_audit['net_payout']:,} SYP")
```

```python
import os
import re

# Selected SQL files for currency audit
sql_files = [
    '/content/model-5-schema-v5.sql',
    '/content/model-2-schema-v3.sql',
    '/content/08_central_accumulator.sql',
    '/content/07_model-7-schema.sql',
    '/content/model-1-schema-v4.sql',
    '/content/model-6-schema-v2.sql',
    '/content/model-3-schema-v3.sql',
    '/content/04_model-4-schema.sql'
]

# Keywords to identify currency and model logic
currency_keywords = ['usd', 'syp', 'ل.س', 'dollar', 'دولار', 'exchange', 'صرف']

def audit_sql_currency(file_path):
    if not os.path.exists(file_path):
        return f"File not found: {file_path}"

    findings = []
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        lines = f.readlines()
        for i, line in enumerate(lines):
            if any(key in line.lower() for key in currency_keywords):
                findings.append(f"Line {i+1}: {line.strip()}")

    return findings

print("--- Currency Logic Audit Results ---\n")
for sql_file in sql_files:
    print(f"FILE: {os.path.basename(sql_file)}")
    results = audit_sql_currency(sql_file)
    if isinstance(results, list) and results:
        for r in results[:10]: # Show top 10 relevant lines
            print(f"  {r}")
        if len(results) > 10:
            print(f"  ... and {len(results)-10} more matches.")
    elif not results:
        print("  No explicit currency keywords found.")
    else:
        print(f"  {results}")
    print("-" * 40)
```

```python
import os
import re

sql_files = [
    '/content/model-5-schema-v5.sql',
    '/content/model-2-schema-v3.sql',
    '/content/08_central_accumulator.sql',
    '/content/07_model-7-schema.sql',
    '/content/model-1-schema-v4.sql',
    '/content/model-6-schema-v2.sql',
    '/content/model-3-schema-v3.sql',
    '/content/04_model-4-schema.sql'
]

currency_patterns = [
    r'.*currency.*',
    r'.*usd.*',
    r'.*syp.*',
    r'.*ل\.س.*',
    r'.*دولار.*',
    r'.*exchange.*',
    r'.*price.*',
    r'.*تومان.*'
]

def extract_currency_logic(file_path):
    if not os.path.exists(file_path):
        return f"File {file_path} not found."

    extracted = []
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read()
        # Look for table definitions or functions containing currency keywords
        # Splitting by semicolon to get individual SQL statements
        statements = content.split(';')
        for stmt in statements:
            if any(re.search(ptrn, stmt, re.IGNORECASE) for ptrn in currency_patterns):
                extracted.append(stmt.strip())
    return extracted

print("=== CURRENCY & FINANCIAL LOGIC EXTRACTION ===\n")
for f_path in sql_files:
    filename = os.path.basename(f_path)
    print(f"ANALYZING: {filename}")
    logic_blocks = extract_currency_logic(f_path)
    if logic_blocks:
        for block in logic_blocks:
            # Print truncated block to keep it readable
            print(f"  [BLOCK FOUND]:\n  {block[:500]}...\n")
    else:
        print("  No explicit currency logic blocks detected.\n")
    print("-" * 60)
```
